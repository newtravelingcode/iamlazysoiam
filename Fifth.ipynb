```python
import pandas as pd
```

```python
df1 = pd.read_csv('2024_가천대.csv', encoding='utf-8')
df2 = pd.read_csv('2025_가천대.csv', encoding='utf-8')
df3 = pd.read_csv('2026_가천대.csv', encoding='utf-8')

df = pd.merge(df1, df2, on=['모집전형', '모집단위'], how='outer', suffixes=('_2024', '_2025'))
df = pd.merge(df, df3, on=['모집전형', '모집단위'], how='outer')

df.info()
df
```

```python
df.to_csv('merged_가천대.csv', index=False, encoding='utf-8-sig')
```
